```python
import requests

# URL for the TCIA Brain Tumor Progression dataset (example link, update if needed)
url = "https://public.cancerimagingarchive.net/tcia/tcia-download-link.zip"

# Path to save the file (replace YOUR_DIRECTORY_PATH)
save_path = "/mnt/Internal/MedImage/Datasets/TCIA"

response = requests.get(url, stream=True)

if response.status_code == 200:
    with open(save_path, 'wb') as file:
        for chunk in response.iter_content(chunk_size=8192):
            file.write(chunk)
    print(f"TCIA Brain Tumor Progression dataset downloaded successfully to {save_path}")
else:
    print(f"Failed to download TCIA dataset. Status code: {response.status_code}")
```
