```python
import os
from IPython.display import display
from PIL import Image
import matplotlib.pyplot as plt

# Path to the folder containing images
# image_folder =  "/mnt/Internal/MedImage/CheXpert Dataset/unzip_chexpert_images/CheXpert-v1.0/train/patient12917/study1"
image_folder = "/mnt/Internal/MedImage/unzip_chexpert_images/CheXpert-v1.0/train/patient21147/study1/"
#image_folder =  "/mnt/Internal/MedImage/CheXpert Dataset/Lab_Rotation_2/LORA_Model_Imagea_generation_with_prompt/CheXpert-v1.0/train/patient00991/study1/"
# List all image files in the folder
images = [f for f in os.listdir(image_folder) if f.endswith(('.png', '.jpg', '.jpeg'))]

# Display images as grayscale
for img_file in images:
    img_path = os.path.join(image_folder, img_file)
    img = Image.open(img_path).convert("L")  # Convert to grayscale
    plt.figure()
    plt.imshow(img, cmap='gray')  # Display as grayscale
    plt.axis('off')  # Hide axes
    plt.show()
```

```python
import os
import shutil

def clear_directory(dir_path):
    for filename in os.listdir(dir_path):
        file_path = os.path.join(dir_path, filename)
        try:
            if os.path.isfile(file_path) or os.path.islink(file_path):
                os.unlink(file_path)  # Delete file
            elif os.path.isdir(file_path):
                shutil.rmtree(file_path)  # Delete folder (if any subfolder exists)
        except Exception as e:
            print(f'Failed to delete {file_path}. Reason: {e}')

# Call the function on your specific path
clear_directory("/mnt/Internal/MedImage/CheXpert Dataset/Lab_Rotation_2/training_our_own_diffusion_model/256_diffusion_model_with_prompt/rough_folder_for_img_checking")
```

```python
import os
import matplotlib.pyplot as plt
import matplotlib.image as mpimg

# Set the path to your images directory
image_dir = "/mnt/Internal/MedImage/CheXpert Dataset/Lab_Rotation_2/training_our_own_diffusion_model/256_diffusion_model_with_prompt/rough_folder_for_img_checking/"

# Specify the disease you're interested in
disease_name = "support devices"  # replace this with your disease of interest

# Get a list of all files in the directory
all_files = os.listdir(image_dir)

# Filter the list to include only images that correspond to the specified disease
# Assuming filenames contain the disease name
disease_images = [f for f in all_files if disease_name in f.lower()]

# Limit to 10 images if there are more than 10
disease_images = disease_images[:10]

# Create a subplot to display the images
fig, axes = plt.subplots(2, 5, figsize=(15, 6))

# Loop through the images and display them
for i, ax in enumerate(axes.flatten()):
    if i < len(disease_images):
        # Load and display the image
        img = mpimg.imread(os.path.join(image_dir, disease_images[i]))
        ax.imshow(img)
        ax.axis('off')  # Hide axes
        ax.set_title(f"Support Devices{i+1}")
    else:
        ax.axis('off')  # Hide axes if fewer than 10 images are available

# Adjust layout to avoid overlapping
plt.tight_layout()
plt.show()
```
