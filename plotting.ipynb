#### Models Comparison - AUC for specific Diseases

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

# Example data (replace this with your actual AUC values)
df_auc = pd.DataFrame({
    "Disease": [
        'No Finding', 'Lung Opacity', 'Lung Lesion',
        'Edema', 'Consolidation', 'Pneumonia', 'Atelectasis', 'Pneumothorax',
        'Pleural Effusion', 'Pleural Other', 'Fracture', 'Support Devices'
    ],
    "DenseNet121 - Real": [0.8159, 0.7148, 0.7245, 0.8098, 0.6721, 0.7606, 0.6812, 0.8569, 0.8625, 0.8048, 0.7894, 0.8806],
    "DenseNet121 - Gen": [0.8036, 0.6736, 0.6873, 0.7695 ,0.6579, 0.7139, 0.651, 0.7412, 0.8104, 0.7096, 0.6565, 0.8076],
    "ResNet50 - Real": [0.849, 0.7142, 0.7603, 0.7974, 0.6579, 0.7372, 0.6792, 0.8113, 0.8377, 0.7354, 0.7276, 0.8495],
    "ResNet50 - Gen": [0.835, 0.699, 0.7326, 0.7726, 0.6508, 0.7167, 0.6574, 0.7464, 0.8101, 0.6799, 0.6817 ,0.8114]
})

# Set up
diseases = df_auc["Disease"]
models = ["DenseNet121 - Real", "DenseNet121 - Gen", "ResNet50 - Real", "ResNet50 - Gen"]
x = np.arange(len(diseases))
width = 0.2

# Plot
plt.figure(figsize=(16, 6))
for i, model in enumerate(models):
    plt.bar(x + i * width, df_auc[model], width=width, label=model)

# Axis settings
plt.xticks(x + width * 1.5, diseases, rotation=45, ha='right')
plt.ylabel("AUC")
plt.title("AUC Comparison Across Diseases by Model")
plt.ylim(0.5, 1.0)
plt.legend(title="Model")
plt.tight_layout()
plt.grid(axis="y", linestyle="--", alpha=0.5)
plt.show()
```
